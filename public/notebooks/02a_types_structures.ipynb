## 1. Numbers: `int` and `float`

`int` = whole numbers (share quantities, number of days). `float` = decimals (prices, returns, ratios).

Rule of thumb in finance: **quantities are ints, money and rates are floats.**

```python
shares_held = 150              # int  - you can't own 150.5 shares of most stocks
price_per_share = 3450.75      # float - rupees and paise
position_value = shares_held * price_per_share

print("Shares:", shares_held, "| type:", type(shares_held).__name__)
print("Price:", price_per_share, "| type:", type(price_per_share).__name__)
print("Position value: Rs", position_value)
```

**⚠️ The float trap every analyst meets:** computers store decimals in binary, so tiny errors creep in. Run this.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)        # False! Surprising but true everywhere, not just Python

# The fix: round before comparing, or compare with a tolerance
print(round(0.1 + 0.2, 10) == 0.3)
```

*Why this matters:* if you reconcile two ledgers with `==` on floats, you will get false mismatches. Real finance systems round or use tolerances.

## 2. Text: `str`

Ticker symbols, company names, categories. Note the `.NS` suffix — that's how NSE-listed stocks are identified on data platforms like yfinance.

```python
ticker = "RELIANCE.NS"
company = "Reliance Industries Limited"

print(ticker.upper())              # already upper, but shows the method
print(ticker.replace(".NS", ""))   # strip the exchange suffix
print(company[:8])                 # first 8 characters - "slicing"
print(len(company), "characters")

# f-strings: the modern way to build text with values inside
qty, px = 150, 3450.75
print(f"Bought {qty} shares of {ticker} at Rs {px:,.2f}")   # :,.2f = comma separator, 2 decimals
```

## 3. Truth: `bool`

`True` / `False`. Every trading rule, every risk limit, every filter is ultimately a bool.

```python
price = 3450.75
stop_loss = 3300.00

is_above_stop = price > stop_loss
print("Above stop loss?", is_above_stop)

# Comparison operators:  >  <  >=  <=  ==  !=
# Logical operators:     and  or  not
large_position = 150 > 100
print("Large position AND above stop?", large_position and is_above_stop)
print("Not above stop?", not is_above_stop)
```

## 4. Lists: ordered collections `[ ]`

Use a list when order matters and items can repeat — a watchlist, a series of daily closes.

```python
watchlist = ["RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS"]

print(watchlist[0])        # first item  - Python counts from 0!
print(watchlist[-1])       # last item
print(watchlist[1:3])      # items 1 and 2 (the end index is NOT included)
print(len(watchlist), "stocks")

watchlist.append("ITC.NS")        # add to the end
watchlist.remove("INFY.NS")       # remove by value
print(watchlist)
```

```python
# Lists of numbers behave like a price series
closes = [3450.75, 3462.10, 3438.90, 3475.25, 3491.00]

print("Latest:", closes[-1])
print("Highest:", max(closes))
print("Lowest:", min(closes))
print("Average:", sum(closes) / len(closes))
```

## 5. Dictionaries: labelled collections `{ }`

**The most important structure in this course.** A dict maps *keys* to *values* — exactly how a portfolio works: ticker → quantity.

```python
portfolio = {
    "RELIANCE.NS": 150,
    "TCS.NS": 80,
    "HDFCBANK.NS": 200,
}

print(portfolio["TCS.NS"])              # look up by key
portfolio["INFY.NS"] = 120              # add a new holding
portfolio["TCS.NS"] = 95                # update an existing one
del portfolio["RELIANCE.NS"]            # sell out completely

print(portfolio)
print("Tickers:", list(portfolio.keys()))
print("Quantities:", list(portfolio.values()))
print("Do we hold ITC?", "ITC.NS" in portfolio)
```

```python
# Two dicts working together - the pattern behind every valuation
holdings = {"TCS.NS": 95, "HDFCBANK.NS": 200, "INFY.NS": 120}
prices   = {"TCS.NS": 3450.75, "HDFCBANK.NS": 1520.40, "INFY.NS": 1480.50}

value_tcs = holdings["TCS.NS"] * prices["TCS.NS"]
print(f"TCS position: Rs {value_tcs:,.2f}")
# In notebook 2B you'll compute the WHOLE portfolio with a loop.
```

## 6. Tuples `( )` and Sets `{ }`

**Tuple** = a list that cannot be changed. Use for fixed records: a single trade, a coordinate.
**Set** = unordered, no duplicates. Use to deduplicate or test membership fast.

```python
# Tuple: one trade, fixed forever once recorded
trade = ("TCS.NS", 50, 3450.75, "2026-08-17")
tkr, qty, px, dt = trade                     # "unpacking"
print(f"{dt}: bought {qty} {tkr} @ {px}")

# trade[1] = 60   # <- uncomment to see TypeError: tuples are immutable

# Set: strip duplicates instantly
traded_today = ["TCS.NS", "INFY.NS", "TCS.NS", "ITC.NS", "INFY.NS", "TCS.NS"]
unique = set(traded_today)
print("Unique tickers traded:", unique, "| count:", len(unique))
```

## Recap

| Type | Syntax | Use in finance |
|---|---|---|
| `int` | `150` | Share quantities, day counts |
| `float` | `3450.75` | Prices, returns, ratios |
| `str` | `"TCS.NS"` | Tickers, names, categories |
| `bool` | `True` | Rules, limits, filters |
| `list` | `[a, b]` | Watchlists, price series (order matters) |
| `dict` | `{k: v}` | **Portfolios**, lookups, mappings |
| `tuple` | `(a, b)` | Fixed records — a completed trade |
| `set` | `{a, b}` | Unique values, fast membership tests |
