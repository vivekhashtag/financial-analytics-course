## Step 3 — Meet your first dataset
Ten days of (sample) NIFTY 50 index levels. In later modules we'll download real data from the internet — for now, we create it right here so nothing can go wrong.

*(Don't worry about understanding this code yet — Modules 2 and 3 will teach every line of it.)*

```python
import pandas as pd

data = {
    "date":  ["2026-01-05", "2026-01-06", "2026-01-07", "2026-01-08", "2026-01-09",
              "2026-01-12", "2026-01-13", "2026-01-14", "2026-01-15", "2026-01-16"],
    "close": [21741.5, 21833.2, 21790.1, 21902.7, 22011.4,
              21955.8, 22102.3, 22187.9, 22140.6, 22251.0],
}

prices = pd.DataFrame(data)
prices["date"] = pd.to_datetime(prices["date"])
prices  # in a notebook, the last line of a cell is displayed automatically
```

**Quick look:** 10 rows, 2 columns — a date and a closing price. Notice there's no Jan 10 or 11: NSE is closed on weekends. (Module 1 taught you: an *expected* gap, not a *problem* gap.)

## Step 4 — Your first chart
If a chart appears below this cell, your environment is 100% ready.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 4))
plt.plot(prices["date"], prices["close"], marker="o")
plt.title("My First Financial Chart — NIFTY 50 Close, Jan 2026")
plt.xlabel("Date")
plt.ylabel("NIFTY 50 closing level")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
