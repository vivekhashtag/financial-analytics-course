## 2B · Control Flow & Functions

```python
# 2B Ex1 - client segmentation
aum = 4_500_000
if aum < 1_000_000:
    segment = "Mass"
elif aum <= 10_000_000:
    segment = "Affluent"
else:
    segment = "HNI"
print(segment)                               # Affluent

# 2B Ex2 - count up days
closes = [3450.75, 3462.10, 3438.90, 3475.25, 3491.00, 3468.30, 3502.15]
up_days = 0
for i in range(1, len(closes)):
    if closes[i] > closes[i - 1]:
        up_days += 1
print("Up days:", up_days)                   # 4

# 2B Ex3 - brokerage with a floor
def brokerage_cost(trade_value, rate_pct=0.03, minimum=20):
    computed = trade_value * rate_pct / 100
    return max(computed, minimum)

print(brokerage_cost(500))                   # 20
print(brokerage_cost(200_000))               # 60.0
```

## 2C · Position Tracker

```python
# Setup (same as 2C)
portfolio = {
    "RELIANCE.NS":  {"qty": 150, "buy_price": 2380.00, "current": 2512.40},
    "TCS.NS":       {"qty":  95, "buy_price": 3620.50, "current": 3450.75},
    "HDFCBANK.NS":  {"qty": 200, "buy_price": 1445.20, "current": 1520.40},
    "INFY.NS":      {"qty": 120, "buy_price": 1530.00, "current": 1480.50},
    "ITC.NS":       {"qty": 500, "buy_price":  398.75, "current":  415.20},
}
def cost_basis(h):   return h["qty"] * h["buy_price"]
def market_value(h): return h["qty"] * h["current"]
def pnl(h):          return market_value(h) - cost_basis(h)
def pnl_pct(h):      return pnl(h) / cost_basis(h) * 100


# 2C Ex1 - add a holding
def add_holding(portfolio, ticker, qty, buy_price, current):
    portfolio[ticker] = {"qty": qty, "buy_price": buy_price, "current": current}

add_holding(portfolio, "SBIN.NS", 300, 610.00, 590.25)
print("Added. Holdings now:", list(portfolio.keys()))


# 2C Ex2 - best and worst by % P&L
def best_and_worst(portfolio):
    ranked = sorted(portfolio.items(), key=lambda pair: pnl_pct(pair[1]), reverse=True)
    return (ranked[0][0], ranked[-1][0])

print(best_and_worst(portfolio))


# 2C Ex3 - sell, removing the holding if it hits zero
def sell(portfolio, ticker, qty):
    if ticker not in portfolio:
        print(f"No position in {ticker}")
        return
    portfolio[ticker]["qty"] -= qty
    if portfolio[ticker]["qty"] <= 0:
        del portfolio[ticker]
        print(f"Closed out {ticker}")
    else:
        print(f"Sold {qty} {ticker}, {portfolio[ticker]['qty']} remaining")

sell(portfolio, "ITC.NS", 100)
sell(portfolio, "INFY.NS", 120)
print("Final holdings:", list(portfolio.keys()))
```
